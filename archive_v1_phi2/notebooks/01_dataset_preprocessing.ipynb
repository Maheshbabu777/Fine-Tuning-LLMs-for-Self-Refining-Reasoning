```python
import numpy
import pandas
import json
```

```python
from pathlib import Path
```

```python
path=Path("../datasets/raw/cot9_dataset.json")
```

```python
with path.open("r",encoding='utf-8') as file:
    dataset=json.load(file)
```

```python
dataset[1]
```

```python
cot_data=[]
dcot_data=[]
modified_dataset=[]
```

```python
modified_dataset=[{'question': i['question'], 'options': i.get('options',None), 'context': i.get('context',None),'correct_cots': i['correct_cots'], 'answer': i['answer']} for i in dataset]
```

```python
modified_dataset[1]
```

```python
cot_data=[{'question': i['question'], 'options': i.get('options',None), 'context': i.get('context',None),'reasoning': j['cot'], 'answer': i['answer']} for i in modified_dataset for j in i['correct_cots']]
```

```python
cot_data[3]
```

```python
len(cot_data)
```

```python
len(dataset)
```

```python
from itertools import permutations
import random
```

```python
for i in modified_dataset:
    for j in range(1,len(i['correct_cots'])+1):
        permut=random.sample(list(permutations(i['correct_cots'],j)),1)[0]
        grouped_cots=[k['cot'] for k in permut]
        dcot_data.append({'question': i['question'], 'options': i.get('options',None), 'context': i.get('context',None),'reasoning': grouped_cots, 'answer': i['answer']})
```

```python
dcot_data[1]
```

```python
len(dcot_data)
```

```python
cot_path=Path('../datasets/processed/cot_intermediate.json')
dcot_path=Path('../datasets/processed/dcot_intermediate.json')
```

```python
with cot_path.open('w',encoding="utf-8") as file:
    json.dump(cot_data,file,ensure_ascii=False,indent=2)

with dcot_path.open('w',encoding="utf-8") as file:
    json.dump(dcot_data,file,ensure_ascii=False,indent=2)
```
