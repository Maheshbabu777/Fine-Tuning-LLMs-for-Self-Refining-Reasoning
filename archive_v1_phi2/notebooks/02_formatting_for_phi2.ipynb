```python
path1=Path('../datasets/processed/cot_intermediate.json')
path2=Path('../datasets/processed/dcot_intermediate.json')
```

```python
with path1.open('r',encoding="utf-8") as file:
    cot_dataset=json.load(file)
with path2.open('r',encoding="utf-8") as file:
    dcot_dataset=json.load(file)
```

```python
def format_for_cot(record: dict)->str:
    result=""

    result+='[Question] '+record.get('question')+'\n'

    if(record.get('options')!=None):
        result+='[Options] '+record.get('options')+'\n'

    if(record.get('context')!=None):
        result+='[Context] '+record.get('context')+'\n'
    
    result+='[Number of answers] '+'1'+'\n'

    result+='[Answer 1] \n'+record.get('reasoning')+'\n'

    result+='[Final answer] '+record['answer']
    
    return result
        
```

```python
def format_for_dcot(record: dict)->str:
    result=""
    
    result+='[Question] '+record.get('question')+'\n'
    
    if(record.get('options')!=None):
        result+='[Options] '+record.get('options')+'\n'
    
    if(record.get('context')!=None):
        result+='[Context] '+record.get('context')+'\n'
        
    result+='[Number of answers] '+str(len(record.get('reasoning')))+'\n'
    
    for i in range(1,len(record.get('reasoning'))+1):
        result+=f'[Answer {i}] \n'+record.get('reasoning')[i-1]+'\n'
    
    result+='[Final answer] '+record['answer']
        
    return result
```
