```python
import json
from pathlib import Path
```

```python
from transformers import AutoTokenizer
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
model_name='microsoft/phi-2'
tokenizer=AutoTokenizer.from_pretrained(model_name)
```

```python
path1=Path('../datasets/processed/formatted_cot.json')
path2=Path('../datasets/processed/formatted_dcot.json')
```

```python
with path1.open('r',encoding='utf-8') as file:
    cot_dataset=json.load(file)

with path2.open('r',encoding='utf-8') as file:
    dcot_dataset=json.load(file)
```

```python
lens_cot=[
    len(tokenizer(cot,truncation=False)['input_ids'])
    for cot in cot_dataset
]

lens_dcot=[
    len(tokenizer(dcot,truncation=False)['input_ids'])
    for dcot in dcot_dataset
]

print('cot')
print('min', min(lens_cot))
print('max', max(lens_cot))
print('avg',sum(lens_cot)/len(lens_cot))


print('dcot')
print('min', min(lens_dcot))
print('max', max(lens_dcot))
print('avg',sum(lens_dcot)/len(lens_dcot))
```

```python
print("CoT > 2048:",sum(x>2048 for x in lens_cot))
print("DCoT > 2048:",sum(x>2048 for x in lens_dcot))
```

```python
cot_trainable=[
    cot_dataset[i] for i in range(len(lens_cot)) if lens_cot[i]<=2048
]

cot_trainable[1]
```

```python
dcot_trainable=[
    dcot_dataset[i] for i in range(len(lens_dcot)) if lens_dcot[i]<=2048
]

dcot_trainable[1]
```

```python
print(len(cot_dataset)-len(cot_trainable))
print(len(dcot_dataset)-len(dcot_trainable))
```

```python
path1=Path('../datasets/phi-2/cot_trainable.json')
path2=Path('../datasets/phi-2/dcot_trainable.json')
```

```python
with path1.open('w',encoding='utf-8') as file:
    json.dump(cot_trainable,file,ensure_ascii=False,indent=2)

with path2.open('w',encoding='utf-8') as file:
    json.dump(dcot_trainable,file,ensure_ascii=False,indent=2)
```
