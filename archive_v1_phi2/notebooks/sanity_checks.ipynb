```python
import numpy as np
import pandas as pd
import json
from pathlib import Path
```

```python
path1=Path('../datasets/phi-2/cot_trainable.json')
path2=Path('../datasets/phi-2/dcot_trainable.json')
```

```python
with path1.open('r',encoding='utf-8') as file:
    cot_dataset=json.load(file)

with path2.open('r',encoding='utf-8') as file:
    dcot_dataset=json.load(file)
```

```python
print(len(cot_dataset))
print(len(dcot_dataset))
```

```python
assert all(isinstance(x,str) and x.strip() for x in cot_dataset)
assert all(isinstance(x,str) and x.strip() for x in dcot_dataset)
```

```python
required_cot=[
    "[Question]",
    "[Number of answers]",
    "[Answer 1]",
    "[Final answer]"
]

for text in cot_dataset:
    for field in required_cot:
        assert field in text , f'Missing: {field}'
```

```python
required_dcot = [
    "[Question]",
    "[Number of answers]",
    "[Answer 1]",
    "[Final answer]"
]

for text in dcot_dataset:
    for field in required_dcot:
        assert field in text, f"Missing {field}"
```

```python
import re
from collections import Counter

dcot_counts=[]

for text in dcot_dataset:
    match=re.search(r"\[Number of answers\]\s+(\d+)",text)
    assert match, "Number of answers not found"
    dcot_counts.append(int(match.group(1)))

print(Counter(dcot_counts))
```
